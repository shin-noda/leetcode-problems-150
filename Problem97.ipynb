```python
class Solution(object):
    def isInterleave(self, s1, s2, s3):
        """
        :type s1: str
        :type s2: str
        :type s3: str
        :rtype: bool
        """

        # Check if s3 can be made the combination of s1 and s2
        if len(s1) + len(s2) != len(s3):
            return False

        # Initialize a 2D boolean array dp of size
        # dp[i][j] = True: means
        # Can the first i character of s1 and the first j characters of s2
        # be interleaved to form the first i + j characters of s3?
        # e.g.
        # s1 = "ab", s2 = "cd", s3 = "acbd"
        # dp[1][1] means:
        # - first 1 character of s1: 'a'
        # - first 1 character of s2: 'c'
        # -> Can they form the first 1 + 1 = 2 characters of s3
        # First 2 characters of s3 are: "ac" -> So yes.

        # Key Insight
        # Let's focus on a single cell: dp[2][1]
        # We want to form the first 3 characters of s3.
        # We have access to s1[:2] and s2[:1].
        # Think about the very last character of that 3 character prefix of s3.
        # Where could that last character have come from?
        # There are only two possibilities:
        # - It came from the end of s1[:2] or
        # - it came from the end of s2[:1].
        #
        # That's why we checkes above and left because
        # those are the two states that could have produced the current one.

        dp = [[False] * (len(s2) + 1) for _ in range(len(s1) + 1)]

        # Set the first element of dp to True
        # This represents two empty strings forming an empty s3,
        # so it's always True
        dp[0][0] = True

        # Fill the first column only using s1
        for i in range(1, len(dp)):
            if dp[i - 1][0] and s1[i - 1] == s3[i - 1]:
                dp[i][0] = True

        # Fill the first row only using s2
        for j in range(1, len(dp[0])):
            if dp[0][j - 1] and s2[j - 1] == s3[j - 1]:
                dp[0][j] = True

        # Fill the rest of the table
        for i in range(1, len(dp)):
            for j in range(1, len(dp[i])):
                # Note: len(s1) + len(s2) = len(s3)
                # The incdices start from 0, so:
                # 1st character:        s3[0]
                # 2nd character:        s3[1]
                # (i + j)-th character: s3[i + j - 1]
                if (
                    dp[i - 1][j] and s1[i - 1] == s3[i + j - 1] or
                    dp[i][j - 1] and s2[j - 1] == s3[i + j - 1]
                ):
                    dp[i][j] = True

        return dp[-1][-1]
```

```python
solution = Solution()
```

```python
s1 = "aabcc"
s2 = "dbbca"
s3 = "aadbbcbcac"

print(solution.isInterleave(s1, s2, s3))
```
